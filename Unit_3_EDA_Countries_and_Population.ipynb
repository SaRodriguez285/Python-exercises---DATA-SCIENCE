## 1. Connection to the course material

A database organizes persistent information. A **DataFrame** is an in-memory tabular structure;
it does not, by itself, replace a database management system. A CSV is a flat file without
the constraints and management features of a relational database system.

The course material describes object-oriented, distributed, warehouse, NoSQL, and relational
databases. Flat tabular files are sufficient for this activity, so no database server is required.
The **data model** must identify the unit of observation and the keys:

| Table | Unit of observation | Verified key | Main variables |
| --- | --- | --- | --- |
| Geography | A country or territory | `alpha_3` | Continent, area, population, and codes |
| Historical population | A country in a particular year | `country`, `year` | Country, year, and population |

These are **secondary data**, because third parties collected them. Continent is a nominal
categorical variable; population is a discrete quantitative variable; area and density are
continuous quantitative variables. Year is an integer-valued temporal indicator.
A numerical identifier such as `geoname_id` is a label, not a measurement to interpret through correlation.

Publicly accessible datasets support reuse and analysis, but their origin, date, license, and
coverage should be checked before using them for decisions. Public access alone does not
guarantee accuracy or up-to-date information. This notebook records its sources and preserves
the exact data snapshots used in the analysis.

## 3. Data loading and structure
**Important parsing detail:** `NA` is both the continent code for North America and the
two-letter country code for Namibia. Pandas normally treats `NA` as a missing value.
We use `keep_default_na=False` and mark only empty fields as missing. Identifier columns
are read as strings to preserve leading zeros and avoid treating codes as measurements.

```python
identifier_types = {
    column: 'string'
    for column in ['alpha_2', 'alpha_3', 'numeric', 'phone', 'fips', 'geoname_id']
}
geography_raw = pd.read_csv(
    StringIO(GEOGRAPHY_CSV), sep=';', keep_default_na=False,
    na_values=[''], dtype=identifier_types,
)
population_raw = pd.read_csv(StringIO(POPULATION_CSV))

for name, table in [('Geography', geography_raw), ('Population', population_raw)]:
    print(f'\n{name}: {table.shape[0]:,} rows and {table.shape[1]} columns')
    display(table.head())
    table.info()

print('Checking a valid NA country code:')
display(geography_raw.loc[geography_raw['alpha_2'] == 'NA', ['name', 'alpha_2', 'continent']])
```

## 4. Data quality assessment
We count missing values, fully duplicated rows, and repeated keys. These are different issues:
two rows with the same key but different values may be conflicting records rather than exact
duplicates. We also check whether area measurements support the later density calculation.

```python
def quality_summary(table):
    return pd.DataFrame({
        'data_type': table.dtypes.astype(str),
        'missing_count': table.isna().sum(),
        'missing_percent': table.isna().mean().mul(100).round(2),
    })

display(quality_summary(geography_raw))
display(quality_summary(population_raw))
print('Fully duplicated geography rows:', geography_raw.duplicated().sum())
print('Fully duplicated population rows:', population_raw.duplicated().sum())
print('Repeated alpha_3 keys:', geography_raw.duplicated(['alpha_3']).sum())
print('Repeated country–year keys:', population_raw.duplicated(['country', 'year']).sum())
print('Records with nonpositive area:')
display(geography_raw.loc[geography_raw['area'] <= 0, ['name', 'area', 'population']])
```

## 5. Data cleaning and manipulation
We work on copies to preserve the raw data. Exact duplicates are removed if present, while
key uniqueness is checked separately. Leading and trailing whitespace is removed from names.
We create population in millions, calculate density in people/km², and label the continents.
Unknown populations are not replaced with zero, and large values are not automatically removed.

Missing values are handled according to their purpose: unused metadata remain missing, while
an undefined density is excluded only from density-specific calculations. This avoids losing
valid population and area observations because of unrelated missing metadata.

```python
geography = geography_raw.drop_duplicates().copy()
population = population_raw.drop_duplicates().copy()
geography['name'] = geography['name'].str.strip()
population['country'] = population['country'].str.strip()

continent_names = {
    'AF': 'Africa', 'AN': 'Antarctica', 'AS': 'Asia', 'EU': 'Europe',
    'NA': 'North America', 'OC': 'Oceania', 'SA': 'South America',
}
geography['continent_name'] = geography['continent'].map(continent_names).astype('category')
geography['population_millions'] = geography['population'] / 1_000_000
# Mask nonpositive areas to avoid dividing by zero; their density remains NaN.
geography['density_people_km2'] = geography['population'] / geography['area'].where(geography['area'] > 0)
population['population_millions'] = population['population'] / 1_000_000

assert geography['alpha_3'].notna().all() and geography['alpha_3'].is_unique
assert not population.duplicated(['country', 'year']).any()
assert population[['country', 'year', 'population']].notna().all().all()
assert (population['population'] > 0).all()
assert (geography['population'] >= 0).all()
assert geography['continent_name'].notna().all()
assert geography.loc[geography['name'] == 'Namibia', 'alpha_2'].iloc[0] == 'NA'
assert geography['density_people_km2'].isna().sum() == 1
print('Key, code, and measurement checks: passed')
print(f'Rows retained: {len(geography):,} geographic and {len(population):,} historical records')
display(geography[['name', 'continent_name', 'area', 'population', 'density_people_km2']].head())
```

**Cleaning outcome:** all original observations are retained in both tables; there were no
duplicates to remove. Only density-specific analyses exclude one row, so they use **251 of
252 records (99.60%)**. Unused metadata are not made artificially complete. Cleaning decisions
should follow the purpose of the analysis rather than aim to remove every null value.

## 6. Descriptive statistics and frequency tables
We summarize meaningful quantitative measurements and exclude identifiers. Because population
is highly skewed, we examine the median, quartiles, and spread alongside the mean.
The continent table reports absolute and relative frequencies. Cumulative frequencies are
used only for ordered density intervals, not for nominal categories such as continents.
Grouping by continent also demonstrates Pandas aggregation and compares median density.

```python
display(geography[['area', 'population', 'density_people_km2']].describe().round(2))

continent_frequency = geography['continent_name'].value_counts().rename_axis('continent').to_frame('frequency')
continent_frequency['percent'] = continent_frequency['frequency'].div(len(geography)).mul(100)
display(continent_frequency.round(2))

continent_summary = geography.groupby('continent_name', observed=True).agg(
    record_count=('alpha_3', 'size'),
    valid_density_count=('density_people_km2', 'count'),
    median_density_people_km2=('density_people_km2', 'median'),
)
display(continent_summary.round(2))

density_intervals = pd.cut(
    geography['density_people_km2'], bins=[0, 10, 50, 100, 500, np.inf],
    labels=['0–10', '>10–50', '>50–100', '>100–500', '>500'], include_lowest=True,
)
density_frequency = density_intervals.value_counts(sort=False).rename_axis('people/km²').to_frame('frequency')
density_frequency['percent'] = density_frequency['frequency'].div(density_frequency['frequency'].sum()).mul(100)
density_frequency['cumulative_frequency'] = density_frequency['frequency'].cumsum()
density_frequency['cumulative_percent'] = density_frequency['percent'].cumsum()
assert density_frequency['frequency'].sum() == 251
display(density_frequency.round(2))

print(f'Geographic population skewness: {geography["population"].skew():.2f}')
display(geography.nlargest(5, 'density_people_km2')[['name', 'density_people_km2']].round(2))
```

**Interpretation:** Africa has 58 records; Europe, 54; Asia, 52; North America, 42; Oceania,
27; South America, 14; and Antarctica, 5. These counts refer to countries and territories,
not people or exclusively sovereign countries. Population skewness is approximately 9.58,
so the mean alone poorly describes the concentration of values. The overall median density
is approximately 78.09 people/km². Monaco is among the most densely populated records.
These figures describe the embedded snapshot, not an up-to-date international ranking.

## 7. Outliers and univariate visualizations
We flag potential outliers using the interquartile-range rule:
`[Q1 − 1.5 × IQR, Q3 + 1.5 × IQR]`. In a very unequal population distribution, a large value
can be genuine. Flagging an outlier is not sufficient reason to delete it.
We retain the observations and compare the original distribution with a logarithmic transformation.

```python
q1, q3 = geography['population'].quantile([0.25, 0.75])
iqr = q3 - q1
lower_bound, upper_bound = q1 - 1.5 * iqr, q3 + 1.5 * iqr
potential_outlier = ~geography['population'].between(lower_bound, upper_bound)
print(f'Potential population outliers identified by IQR: {int(potential_outlier.sum())}')
display(geography.loc[potential_outlier, ['name', 'population']].sort_values('population', ascending=False).head(10))

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
continent_frequency['frequency'].sort_values().plot.barh(ax=axes[0], color='#0072B2')
axes[0].set(title='Countries and territories by continent', xlabel='Records', ylabel='Continent')
axes[1].hist(geography['population_millions'], bins=25, color='#009E73', edgecolor='white')
axes[1].set(title='Population: original scale', xlabel='Population (millions)', ylabel='Records')
axes[2].hist(np.log10(geography['population'] + 1), bins=25, color='#D55E00', edgecolor='white')
axes[2].set(title='Population: transformed scale', xlabel='log10(population + 1)', ylabel='Records')
fig.tight_layout()
plt.show()
```

**Interpretation:** very large populations create a long right tail and compress most records
into the left-hand portion of the original-scale histogram. The transformation
`log10(population + 1)` makes smaller values easier to examine and accommodates zero populations.
It does not alter the source data or establish normality. The IQR rule flags 33 potential
outliers; all are retained because there is no evidence that their populations are erroneous.

## 8. Relationships between numerical variables
We calculate Pearson's correlation between area and population, excluding identifiers.
A log-scale scatter plot makes very different magnitudes easier to compare. Zero values
are omitted only from that plot because the logarithm of zero is undefined.
The reported correlation uses the original scale; it is not a correlation of log-transformed values.

```python
correlation = geography[['area', 'population']].corr()
display(correlation.round(3))
positive_records = geography.loc[(geography['area'] > 0) & (geography['population'] > 0)]
print(f'Records in the log-scale scatter plot: {len(positive_records)} of {len(geography)}')

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.scatter(positive_records['area'], positive_records['population'], alpha=0.6, color='#0072B2')
ax.set_xscale('log')
ax.set_yscale('log')
ax.set(title='Area and population: countries and territories', xlabel='Area (km², log scale)',
       ylabel='Population (people, log scale)')
fig.tight_layout()
plt.show()
```

**Interpretation:** Pearson's coefficient is approximately **0.413**, indicating a positive
association with considerable variation remaining. A larger area does not, by itself,
determine a larger population. Extreme values can influence Pearson's coefficient.
Linearity, normality, and constant variance have not been established for statistical inference.
Correlation alone does not establish causation or justify automatically removing a variable.

## 9. Contingency table and categorical association
As an additional application of the course material, we cross-tabulate continent and density
**below the median** versus **at or above the median**. This classification is relative to
this dataset, not an official demographic standard. We calculate row percentages, chi-square,
Pearson's contingency coefficient, and Cramér's V.

Before interpreting chi-square, we check the expected frequencies: none should be below 1,
and no more than 20% should be below 5. These numerical checks do not establish independence
or turn the records into a random sample.

```python
density_data = geography.dropna(subset=['density_people_km2']).copy()
median_density = density_data['density_people_km2'].median()
density_data['density_level'] = np.where(
    density_data['density_people_km2'] >= median_density, 'At or above median', 'Below median',
)
contingency = pd.crosstab(density_data['continent_name'], density_data['density_level'])
display(contingency)
display(contingency.div(contingency.sum(axis=1), axis=0).mul(100).round(2))

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(contingency)
n = contingency.to_numpy().sum()
m = min(contingency.shape[0] - 1, contingency.shape[1] - 1)
cramers_v = np.sqrt(chi2 / (n * m))
contingency_coefficient = np.sqrt(chi2 / (chi2 + n))
fraction_below_5 = (expected < 5).mean()
expected_count_conditions = (expected >= 1).all() and fraction_below_5 <= 0.20
print(f'Median threshold: {median_density:.2f} people/km²')
print(f'Minimum expected frequency: {expected.min():.2f}; expected cells below 5: {fraction_below_5:.2%}')
print(f'Numerical expected-frequency conditions met: {expected_count_conditions}')
print(f'Chi-square: {chi2:.3f}; df: {degrees_of_freedom}; calculated p-value: {p_value:.3g}')
print(f"Cramér's V: {cramers_v:.3f}; contingency coefficient: {contingency_coefficient:.3f}")
```

**Interpretation and scope:** Cramér's V is approximately **0.420**, indicating descriptive
association between continent and the chosen density classification. The expected-frequency
conditions are met: the minimum is approximately 2.49 and 14.29% of cells are below 5.
Although the calculated p-value is small, these records do not come from a documented random
sample and may have geographic dependencies. The calculation is therefore presented as an
exercise, **not as evidence for population-wide generalization or causation**.
Phi is not interpreted because the contingency table is not 2 × 2.

## 10. Historical population: Colombia and Argentina
We filter the two countries and align their records **by year** using `pivot`.
Matching records by row position would be unsafe if their order differed or a year were missing.
Only the historical table is used here; its observations are not replaced with geographic-table populations.

```python
selected_countries = population.loc[population['country'].isin(['Colombia', 'Argentina'])].copy()
comparison = selected_countries.pivot(index='year', columns='country', values='population').sort_index()
comparison = comparison[['Colombia', 'Argentina']]
assert comparison.shape == (12, 2)
assert not comparison.isna().any().any(), 'A country is missing in at least one year; review before comparing'
display(comparison)
display(comparison.describe().round(2))

# Each change compares consecutive observations five years apart, not consecutive years.
interval_growth = comparison.pct_change(fill_method=None).mul(100)
print('Percentage change from the previous observation (five-year intervals):')
display(interval_growth.round(2))

first_year, last_year = int(comparison.index.min()), int(comparison.index.max())
elapsed_years = last_year - first_year
growth_summary = pd.DataFrame({
    'initial_population': comparison.iloc[0],
    'final_population': comparison.iloc[-1],
    'increase_people': comparison.iloc[-1] - comparison.iloc[0],
    'total_growth_percent': (comparison.iloc[-1] / comparison.iloc[0] - 1) * 100,
    'compound_annual_growth_percent': ((comparison.iloc[-1] / comparison.iloc[0]) ** (1 / elapsed_years) - 1) * 100,
})
display(growth_summary.round(2))
print('First observed year with a larger population in Colombia than in Argentina:',
      comparison.index[comparison['Colombia'] > comparison['Argentina']].min())
```

**Interpretation:** Colombia increases from **12,350,771** people in 1952 to **44,227,550**
in 2007, approximately **258.10%** total growth and a **2.35%** compound annual growth rate.
Argentina increases from **17,876,956** to **40,301,927**, approximately **125.44%** total
growth and a **1.49%** compound annual growth rate.

**1992** is the first observed year in which Colombia exceeds Argentina. Five-year observations
cannot establish the exact year of the crossover. The first `pct_change` result is missing
because no earlier observation exists; this is expected and should not be filled with zero.
These interval changes are five-year percentages, not annual growth rates.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
(comparison / 1_000_000).plot(ax=axes[0], marker='o', color=['#0072B2', '#D55E00'])
axes[0].set(title='Population over time, 1952–2007', xlabel='Year', ylabel='Population (millions)')

base_100_index = comparison.div(comparison.iloc[0]).mul(100)
base_100_index.plot(ax=axes[1], marker='o', color=['#0072B2', '#D55E00'])
axes[1].set(title='Relative growth: 1952 = 100', xlabel='Year', ylabel='Population index')
for ax in axes:
    ax.grid(alpha=0.25)
    ax.legend(title='Country')
fig.tight_layout()
plt.show()
```

**Reading the charts:** the left panel compares population sizes; the right panel removes
the initial difference in scale by setting both countries to a baseline of 100.
Colombia has greater relative population growth during the period. Lines between points
help visualize the trajectories but do not represent additional annual measurements.

## 11. Descriptive regression and assumption review
The course material introduces regression. As an additional demonstration, we fit a straight
line to Colombia's population over time. Year is centered on 1952 so the slope is measured
in millions of people per year. Residuals are examined rather than assuming that a straight
line fully describes the process.

This fit summarizes observed values. It is not used for forecasting, hypothesis testing, or
causal explanation: time series may contain autocorrelation and nonlinear trends.

```python
x = comparison.index.to_numpy() - first_year
y = comparison['Colombia'].to_numpy() / 1_000_000
slope, intercept = np.polyfit(x, y, deg=1)
fitted = slope * x + intercept
residuals = y - fitted
r_squared = 1 - np.sum(residuals ** 2) / np.sum((y - y.mean()) ** 2)
print(f'Fitted population (millions) = {intercept:.3f} + {slope:.3f} × (year − {first_year})')
print(f'Descriptive in-sample R²: {r_squared:.3f}')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(comparison.index, y, 'o', label='Observed data', color='#0072B2')
axes[0].plot(comparison.index, fitted, label='Fitted line', color='#D55E00')
axes[0].set(title='Colombia: descriptive linear fit', xlabel='Year', ylabel='Population (millions)')
axes[0].legend()
axes[1].scatter(comparison.index, residuals, color='#0072B2')
axes[1].axhline(0, color='black', linestyle='--')
axes[1].set(title='Residuals of the linear fit', xlabel='Year', ylabel='Observed − fitted (millions)')
fig.tight_layout()
plt.show()
```

**Interpretation:** a high in-sample R² does not guarantee reliable forecasts. The residuals
show a curved pattern, so a straight line does not fully capture the trajectory.
Normality, constant variance, and independence of errors are not established by this plot.
The absolute slope is not an annual percentage growth rate; compound annual growth was
calculated separately in Section 10.
